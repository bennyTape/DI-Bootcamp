## ⚙️ Configuration <a name="config"></a>

Ce notebook peut fonctionner de deux façons :

- **Mode API réel** : si vous disposez d'une clé API (OpenAI, Gemini, DeepSeek...), vous pouvez l'utiliser pour générer de vraies réponses au fil des étapes.
- **Mode simulation (fallback)** : si aucune clé API n'est disponible ou si le réseau n'est pas accessible (ex. environnement Colab restreint), le notebook utilise des réponses simulées, rédigées manuellement pour illustrer un résultat réaliste et permettre l'évaluation.

Le notebook détecte automatiquement quel mode utiliser.


```python
# Installation (au besoin) et détection du mode d'exécution
import os
import json
import textwrap

USE_REAL_API = False  # Passez à True si vous configurez une clé API valide ci-dessous

# Exemple de configuration pour l'API OpenAI (facultatif) :
# os.environ["OPENAI_API_KEY"] = "votre_cle_ici"
# USE_REAL_API = True

if USE_REAL_API and os.environ.get("OPENAI_API_KEY"):
    try:
        from openai import OpenAI
        client = OpenAI()
        print("✅ Mode API réel activé.")
    except Exception as e:
        print("⚠️ Impossible d'initialiser le client API, bascule en mode simulation.", e)
        USE_REAL_API = False
else:
    print("🧪 Mode simulation activé : les réponses affichées sont pré-rédigées pour illustrer le résultat attendu.")
```

```python
def generate(prompt, simulated_response=None, model="gpt-4"):
    """
    Génère une réponse à partir d'un prompt.
    - Si USE_REAL_API est True et qu'un client est configuré, appelle réellement le modèle.
    - Sinon, retourne la réponse simulée fournie (mode fallback), pour garder le notebook
      exécutable de bout en bout sans dépendance réseau ou clé API.
    """
    if USE_REAL_API and os.environ.get("OPENAI_API_KEY"):
        response = client.chat.completions.create(
            model=model,
            messages=[{"role": "user", "content": prompt}],
            max_tokens=300,
        )
        return response.choices[0].message.content
    else:
        return simulated_response

print("Fonction generate() prête à l'emploi.")
```

## 📄 Texte source (politique d'accès distant)

Voici le paragraphe de politique interne à transformer tout au long de ce notebook.

```python
source_text = (
    "Employees must ensure that all remote access to internal systems is established "
    "via the approved secure VPN. Under no circumstances should unsecured connections "
    "or personal devices lacking endpoint protection be used to access proprietary "
    "data or sensitive communications."
)

print(textwrap.fill(source_text, width=90))
```

## ✍️ Étape 1 — Prompt maîtrisant ton, structure et longueur <a name="etape1"></a>

**Contraintes demandées :**
- Ton amical et clair
- Format en puces (bullet points)
- Paraphrase uniquement (pas de citation directe)
- Moins de 75 mots au total


```python
prompt_1 = f"""
Tu es un rédacteur pédagogique interne. Reformule le texte ci-dessous en français, dans un ton
amical et accessible, destiné à des employés non techniques d'une newsletter interne.

Contraintes :
- Utilise des puces (bullet points), 3 à 4 maximum
- Paraphrase entièrement le contenu : n'utilise aucune citation directe du texte original
- Reste sous la barre des 75 mots au total
- Ne mentionne aucune information, technologie ou recommandation qui n'est pas déjà présente dans le texte source

Texte source :
\"\"\"{source_text}\"\"\"
"""

print(prompt_1)
```

```python
simulated_output_1 = """
🔐 Accès à distance : ce qu'il faut retenir
- Connecte-toi toujours aux systèmes de l'entreprise via le VPN sécurisé officiel.
- Évite les connexions non sécurisées, où que tu sois.
- N'utilise jamais un appareil personnel non protégé pour consulter des données
  sensibles ou des échanges confidentiels.
"""

output_1 = generate(prompt_1, simulated_response=simulated_output_1)
print(output_1)
print(f"Nombre de mots : {len(output_1.split())}")
```

## 🚨 Étape 3 — Détection et mitigation des hallucinations <a name="etape3"></a>

Si le modèle ajoute des éléments absents du texte source (nouvelles technologies, nouvelles règles...),
on renforce le prompt pour le contraindre strictement au contenu d'entrée.

```python
prompt_1_revised = f"""
Tu es un rédacteur pédagogique interne. Reformule le texte ci-dessous en français, dans un ton
amical et accessible, destiné à des employés non techniques d'une newsletter interne.

Contraintes :
- Utilise des puces (bullet points), 3 à 4 maximum
- Paraphrase entièrement le contenu : n'utilise aucune citation directe du texte original
- Reste sous la barre des 75 mots au total
- IMPORTANT : paraphrase uniquement le contenu du texte fourni. N'ajoute aucune nouvelle
  recommandation, technologie, règle ou exemple qui ne figure pas déjà dans le texte source.
- Si une information n'est pas explicitement présente dans le texte source, ne l'invente pas.

Texte source :
\"\"\"{source_text}\"\"\"
"""

print(prompt_1_revised)
```

```python
simulated_output_1_revised = """
🔐 Accès à distance : ce qu'il faut retenir
- Utilise uniquement le VPN sécurisé approuvé pour te connecter aux systèmes internes.
- N'utilise jamais une connexion non sécurisée pour accéder aux données de l'entreprise.
- Les appareils personnels sans protection ne doivent pas servir à consulter des
  données confidentielles ou des échanges sensibles.
"""

output_1_revised = generate(prompt_1_revised, simulated_response=simulated_output_1_revised)
print(output_1_revised)
print(f"Nombre de mots : {len(output_1_revised.split())}")
```

**Comparaison avant / après mitigation :**

| Aspect | Avant (Étape 1) | Après (Étape 3) |
|---|---|---|
| Risque d'ajout de contenu | Le modèle pourrait ajouter des conseils génériques (ex. "activez le MFA") non présents dans le texte source | Contrainte explicite empêchant toute invention d'information |
| Fidélité au texte source | Bonne, mais non garantie | Renforcée par une instruction explicite anti-hallucination |
| Fiabilité pour diffusion officielle | Moyenne | Élevée |

💡 **Bonne pratique repérée :** ajouter une instruction explicite du type *"N'ajoute aucune nouvelle recommandation, technologie ou règle qui ne figure pas dans le texte source"* est une technique simple et efficace pour réduire les hallucinations dans les tâches de paraphrase.


## 🎓 Étape 4 — Paraphrase pour un public junior (stagiaire) <a name="etape4"></a>

**Contraintes demandées :**
- Maximum 4 puces
- Aucun jargon corporate/juridique
- Ton bienveillant et pédagogique


```python
prompt_2_intern = f"""
Tu expliques une règle de sécurité informatique à un(e) stagiaire qui découvre l'entreprise.
Reformule le texte ci-dessous en français simple, avec des phrases courtes, sans aucun jargon
juridique ou corporate.

Contraintes :
- 4 puces maximum
- Vocabulaire simple, accessible à quelqu'un qui n'a jamais travaillé en entreprise
- Ton chaleureux, encourageant et pédagogique (comme si tu accueillais la personne)
- Paraphrase uniquement le contenu du texte source, sans rien inventer

Texte source :
\"\"\"{source_text}\"\"\"
"""

print(prompt_2_intern)
```

```python
simulated_output_2_intern = """
👋 Petit guide sécurité pour bien démarrer
- Pour te connecter aux outils de l'entreprise à distance, utilise toujours le VPN
  officiel : c'est ta connexion sécurisée.
- Évite le wifi public ou toute connexion "libre" pour accéder aux outils du travail.
- Si tu utilises ton propre ordinateur ou téléphone, vérifie qu'il est bien protégé
  avant de t'en servir pour le travail.
- Pas d'inquiétude : en cas de doute, demande toujours à l'équipe IT, on est là pour t'aider !
"""

output_2_intern = generate(prompt_2_intern, simulated_response=simulated_output_2_intern)
print(output_2_intern)
```

## 💬 Étape 5 — Extraction de citation directe <a name="etape5"></a>

Contrairement aux étapes précédentes, il s'agit ici d'extraire **une citation directe** (et non
une paraphrase) qui capture le mieux la politique de sécurité centrale.

```python
prompt_3_quote = f"""
Voici un extrait d'une politique interne. Extrais UNE SEULE citation directe (mot pour mot,
entre guillemets) qui capture le mieux la règle de sécurité centrale du texte. Ne modifie pas
le texte cité et ne fournis aucune paraphrase, seulement la citation exacte.

Texte source :
\"\"\"{source_text}\"\"\"
"""

print(prompt_3_quote)
```

```python
simulated_output_3_quote = (
    '"Employees must ensure that all remote access to internal systems is established '
    'via the approved secure VPN."'
)

output_3_quote = generate(prompt_3_quote, simulated_response=simulated_output_3_quote)
print(output_3_quote)
```

### 🤔 Réflexion

**Dans quel type de communication interne la citation directe serait-elle plus appropriée que la paraphrase ?**

La citation directe est préférable lorsque la formulation exacte a une valeur légale, contractuelle ou
disciplinaire — par exemple dans un mémo de conformité, un rappel de politique officielle envoyé par
le service juridique ou RH, ou un document servant de référence en cas d'audit. Dans ces contextes, la
moindre reformulation pourrait modifier la portée ou l'interprétation de la règle.

**Quand la citation peut-elle poser un risque ?**

- Si le texte source est trop dense ou jargonneux, une citation brute peut nuire à la compréhension
  du public visé (ex. nouveaux employés, stagiaires).
- Une citation sortie de son contexte peut être mal interprétée si elle est diffusée isolément
  (ex. sur un canal de communication informel comme un chat d'équipe).
- Citer un texte protégé ou confidentiel dans un canal externe peut poser un risque de conformité
  ou de fuite d'information si la diffusion n'est pas maîtrisée.

- Le choix entre citation et paraphrase dépend du contexte de diffusion et de l'enjeu de fidélité au texte source.
