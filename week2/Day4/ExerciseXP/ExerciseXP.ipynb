Exercice 1 : Comprendre la visualisation des données
Pourquoi la visualisation est importante :
La visualisation des données permet de transformer des données brutes complexes en représentations graphiques compréhensibles. Elle facilite la détection de tendances, anomalies et patterns, accélère la prise de décision et rend l'information accessible à un public non technique.
Utilité d'un graphique linéaire :
Un graphique linéaire est idéal pour représenter l'évolution d'une variable dans le temps. Il permet de visualiser clairement les tendances, les hausses, les baisses et les cycles sur une période donnée.

```python
# Exercise 2

import matplotlib.pyplot as plt

jours = ["Lun", "Mar", "Mer", "Jeu", "Ven", "Sam", "Dim"]
temperatures = [72, 74, 76, 80, 82, 78, 75]

plt.plot(jours, temperatures, marker='o', color='steelblue', linewidth=2)
plt.xlabel("Jour")
plt.ylabel("Température (°F)")
plt.title("Variation de température sur une semaine")
plt.tight_layout()
plt.show()
```

```python
# Exercise 3
import matplotlib.pyplot as plt

mois = ["Jan", "Fév", "Mar", "Avr", "Mai"]
ventes = [5000, 5500, 6200, 7000, 7500]

plt.bar(mois, ventes, color='coral')
plt.xlabel("Mois")
plt.ylabel("Montant des ventes ($)")
plt.title("Ventes mensuelles d'un magasin de détail")
plt.tight_layout()
plt.show()
```

```python
# Exercise 4

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Student Mental health.csv")

sns.histplot(data=df, x="What is your CGPA?", color="mediumpurple", shrink=0.8)
plt.title("Distribution des moyennes cumulatives (CGPA)")
plt.xlabel("CGPA")
plt.ylabel("Nombre d'étudiants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Exercise 5

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Student Mental health.csv")

sns.countplot(
    data=df,
    x="Choose your gender",
    hue="Do you have Anxiety?",
    palette="Set2"
)
plt.title("Niveaux d'anxiété selon le sexe")
plt.xlabel("Sexe")
plt.ylabel("Nombre d'étudiants")
plt.legend(title="Anxiété")
plt.tight_layout()
plt.show()
```

```python
# Exercise 6

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Student Mental health.csv")

# Conversion Oui/Non → 1/0
df["Panic_numeric"] = df["Do you have Panic attack?"].map({"Yes": 1, "No": 0})

sns.scatterplot(
    data=df,
    x="Age",
    y="Panic_numeric",
    hue="Do you have Panic attack?",
    palette={"Yes": "tomato", "No": "steelblue"},
    s=80,
    alpha=0.7
)
plt.title("Relation entre l'âge et les crises de panique")
plt.xlabel("Âge")
plt.ylabel("Crise de panique (1 = Oui, 0 = Non)")
plt.legend(title="Crise de panique")
plt.tight_layout()
plt.show()
```
