```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import norm, binom
 
# ──────────────────────────────────────────────
# Exercise 1 – Basic Usage of SciPy
# ──────────────────────────────────────────────
import scipy
print("=" * 50)
print("Exercise 1: SciPy Version")
print("=" * 50)
print(f"SciPy version: {scipy.__version__}")
 
```

```python
# ──────────────────────────────────────────────
# Exercise 2 – Descriptive Statistics
# ──────────────────────────────────────────────
print("\n" + "=" * 50)
print("Exercise 2: Descriptive Statistics")
print("=" * 50)
 
data = [12, 15, 13, 12, 18, 20, 22, 21]
 
print(f"Data       : {data}")
print(f"Mean       : {np.mean(data):.2f}")
print(f"Median     : {np.median(data):.2f}")
print(f"Variance   : {np.var(data, ddof=1):.2f}")   # ddof=1 → sample variance
print(f"Std Dev    : {np.std(data, ddof=1):.2f}")
```

```python
# ──────────────────────────────────────────────
# Exercise 3 – Normal Distribution Plot
# ──────────────────────────────────────────────
print("\n" + "=" * 50)
print("Exercise 3: Normal Distribution (mean=50, std=10)")
print("=" * 50)
 
x   = np.linspace(10, 90, 300)
pdf = norm.pdf(x, loc=50, scale=10)
 
plt.figure(figsize=(7, 4))
plt.plot(x, pdf, color="steelblue", linewidth=2)
plt.fill_between(x, pdf, alpha=0.25, color="steelblue")
plt.title("Normal Distribution  (μ=50, σ=10)")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.tight_layout()
plt.savefig("ex3_normal_distribution.png", dpi=120)
plt.close()
print("Plot saved → ex3_normal_distribution.png")
```

```python
# ──────────────────────────────────────────────
# Exercise 4 – T-Test
# ──────────────────────────────────────────────
print("\n" + "=" * 50)
print("Exercise 4: T-Test")
print("=" * 50)
 
np.random.seed(42)
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)
 
t_stat, p_value = stats.ttest_ind(data1, data2)
print(f"T-statistic : {t_stat:.4f}")
print(f"P-value     : {p_value:.4f}")
print("Conclusion  :", "Significant difference (p < 0.05)" if p_value < 0.05
                       else "No significant difference (p ≥ 0.05)")
```

```python
# ──────────────────────────────────────────────
# Exercise 5 – Linear Regression
# ──────────────────────────────────────────────
print("\n" + "=" * 50)
print("Exercise 5: Linear Regression – Housing Prices")
print("=" * 50)
 
house_sizes  = [50, 70, 80, 100, 120]
house_prices = [150_000, 200_000, 210_000, 250_000, 280_000]
 
slope, intercept, r_value, p_value, std_err = stats.linregress(
    house_sizes, house_prices
)
 
print(f"Slope       : {slope:.2f}  (each extra m² adds ~{slope:.0f} currency units)")
print(f"Intercept   : {intercept:.2f}")
print(f"R²          : {r_value**2:.4f}")
 
size_to_predict = 90
predicted_price = slope * size_to_predict + intercept
print(f"Predicted price for 90 m²: {predicted_price:,.0f} currency units")
print("Interpretation: the slope means that for every additional square metre,")
print(f"  the house price increases by approximately {slope:.0f} currency units.")
```

```python
# ──────────────────────────────────────────────
# Exercise 6 – ANOVA
# ──────────────────────────────────────────────
print("\n" + "=" * 50)
print("Exercise 6: One-Way ANOVA – Fertilizers")
print("=" * 50)
 
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]
 
f_stat, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)
print(f"F-value     : {f_stat:.4f}")
print(f"P-value     : {p_value:.4f}")
 
if p_value < 0.05:
    print("Conclusion  : Fertilizers have significantly different effects (p < 0.05).")
else:
    print("Conclusion  : No significant difference between fertilizers (p ≥ 0.05).")
 
print("Note: If p > 0.05 we would fail to reject H₀, meaning we cannot conclude")
print("  that any fertilizer produces meaningfully different plant growth.")
```

```python
 #──────────────────────────────────────────────
# Exercise 7 – Binomial Distribution (Optional)
# ──────────────────────────────────────────────
print("\n" + "=" * 50)
print("Exercise 7: Binomial Distribution – Coin Flips")
print("=" * 50)
 
n, p = 10, 0.5   # 10 flips, fair coin
k    = 5         # exactly 5 heads
 
prob_exactly_5 = binom.pmf(k, n, p)
print(f"P(exactly 5 heads in 10 flips) = {prob_exactly_5:.4f}")
 
# Show full distribution
print("\nFull distribution:")
for heads in range(n + 1):
    bar = "█" * int(binom.pmf(heads, n, p) * 200)
    print(f"  {heads:2d} heads: {binom.pmf(heads, n, p):.4f}  {bar}")
```

```python
# ──────────────────────────────────────────────
# Exercise 8 – Correlation Coefficients (Optional)
# ──────────────────────────────────────────────
print("\n" + "=" * 50)
print("Exercise 8: Pearson & Spearman Correlation")
print("=" * 50)
 
df = pd.DataFrame({
    "age":    [23, 25, 30, 35, 40],
    "income": [35_000, 40_000, 50_000, 60_000, 70_000]
})
 
pearson_r,  pearson_p  = stats.pearsonr(df["age"], df["income"])
spearman_r, spearman_p = stats.spearmanr(df["age"], df["income"])
 
print(f"Pearson  r = {pearson_r:.4f}  (p={pearson_p:.4f})")
print(f"Spearman r = {spearman_r:.4f}  (p={spearman_p:.4f})")
print("Both coefficients are close to 1 → strong positive linear relationship")
print("between age and income in this sample.")
 
print("\n  All exercises completed.")
```
