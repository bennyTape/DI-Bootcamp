Columns such as **State/Region**, **Gender**, and **Age group** are **qualitative** because they describe categories or labels rather than measurable amounts.

For the **Global Trends in Mental Health Disorder** dataset, **Year** and variables such as **Depression prevalence**, **Anxiety rate**, or **Disorder percentage** are **quantitative** because they represent measurable numerical values. Columns such as **Country** and **Disorder type** are **qualitative** because they classify data into groups or categories.

For the **Credit Card Approvals** dataset, **Age**, **Income**, **Credit score**, and **Debt** are **quantitative** because they contain measurable numerical information. Exercise 6: Identifying Data Types
Below are various data sources. Identify whether each one is an example of structured or unstructured data.

A company’s financial reports stored in an Excel file.
Photographs uploaded to a social media platform.
A collection of news articles on a website.
Inventory data in a relational database.
Recorded interviews from a market research study.



1. **A company’s financial reports stored in an Excel file** → **Structured data** because the information is organized into rows and columns with a predefined format.

2. **Photographs uploaded to a social media platform** → **Unstructured data** because images do not follow a fixed tabular structure and require processing techniques to extract information.

3. **A collection of news articles on a website** → **Unstructured data** because the text content does not follow a strict predefined format and can vary in length and structure.

4. **Inventory data in a relational database** → **Structured data** because relational databases store information in organized tables with defined fields and relationships.

5. **Recorded interviews from a market research study** → **Unstructured data** because audio recordings do not have a predefined organization and usually need transcription or analysis before processing.


 Exercise 7: Transformation Exercise
For each of the following unstructured data sources, propose a method to convert it into structured data. Explain your reasoning.

A series of blog posts about travel experiences.
Audio recordings of customer service calls.
Handwritten notes from a brainstorming session.
A video tutorial on cooking.



For a **series of blog posts about travel experiences**, the data can be converted into structured form using **text mining or Natural Language Processing (NLP)**. Key information such as destination, travel date, cost, activities, and sentiment can be extracted and stored in a table. This is useful because blog posts are unstructured text, but they often contain repeated patterns of useful information that can be standardized into columns.

For **audio recordings of customer service calls**, the first step is **speech-to-text transcription** to convert audio into written text. Then NLP techniques can be used to extract structured information such as customer issue type, call duration, resolution status, and customer satisfaction. This transformation is necessary because audio data cannot be directly analyzed in tabular form.

For **handwritten notes from a brainstorming session**, the data can be converted using **OCR (Optical Character Recognition)** to digitize the handwritten text. After that, keyword extraction or manual categorization can be used to organize ideas into structured fields such as topic, idea category, priority level, and contributor. This works because handwritten notes are free-form but usually contain separable ideas.

For **a video tutorial on cooking**, the data can be structured by using **video analysis and transcription tools**. Speech can be converted into text, and visual steps can be labeled manually or automatically to extract structured steps such as ingredients, cooking time, temperature, and instructions. This is useful because videos combine audio and visual unstructured data, but the cooking process itself follows a clear sequence that can be organized.
