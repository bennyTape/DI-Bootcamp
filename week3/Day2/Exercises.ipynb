```python
#Exersercice 1

import numpy as np

# Création d'une matrice 3x3
A = np.array([
    [2, 1, 3],
    [0, 4, 1],
    [5, 2, 8]
])

print("=" * 40)
print("       OPÉRATIONS MATRICIELLES")
print("=" * 40)

print("\n Matrice A :")
print(A)

# 1. Calcul du déterminant
det = np.linalg.det(A)
print(f"\n1️  Déterminant de A : {det:.4f}")

# Vérification si la matrice est inversible
if abs(det) < 1e-10:
    print("  La matrice n'est pas inversible (déterminant ≈ 0).")
else:
    print(" La matrice est inversible (déterminant ≠ 0).")

    # 2. Calcul de l'inverse
    A_inv = np.linalg.inv(A)
    print("\n2️  Inverse de A :")
    print(np.round(A_inv, 4))

    # Vérification : A * A_inv doit donner la matrice identité
    identite = np.round(A @ A_inv, 10)
    print("\n Vérification — A × A⁻¹ (doit être la matrice identité) :")
    print(identite)

print("=" * 40)
```

```python
#EXERCICE 6 

np.random.seed(42)

# Données fournies
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after  = productivity_before + np.random.normal(loc=5, scale=3, size=30)

print("=" * 52)
print("        TEST D'HYPOTHÈSES — NUMPY")
print("=" * 52)

# ─────────────────────────────────────────
# FORMULATION DE L'HYPOTHÈSE
# ─────────────────────────────────────────
print("""
 Hypothèses :
   H0 (nulle)      : La formation n'a aucun effet
                     → moyenne de la différence = 0
   H1 (alternative): La formation améliore la productivité
                     → moyenne de la différence > 0
""")

# ─────────────────────────────────────────
# STATISTIQUES DESCRIPTIVES
# ─────────────────────────────────────────
print("─" * 52)
print("1️  STATISTIQUES DESCRIPTIVES")
print("─" * 52)
print(f"  Moyenne avant  : {np.mean(productivity_before):.2f}")
print(f"  Moyenne après  : {np.mean(productivity_after):.2f}")

differences = productivity_after - productivity_before
print(f"\n  Différence moyenne       : {np.mean(differences):.2f}")
print(f"  Écart type des différences: {np.std(differences):.2f}")

# ─────────────────────────────────────────
# TEST MANUEL (t-test apparié simplifié)
# ─────────────────────────────────────────
print("\n" + "─" * 52)
print("2️  CALCUL DE LA STATISTIQUE t")
print("─" * 52)

n           = len(differences)
moy_diff    = np.mean(differences)
std_diff    = np.std(differences, ddof=1)   # ddof=1 → écart type de l'échantillon
erreur_std  = std_diff / np.sqrt(n)
t_stat      = moy_diff / erreur_std

print(f"  Taille échantillon (n) : {n}")
print(f"  Moyenne des différences: {moy_diff:.2f}")
print(f"  Écart type             : {std_diff:.2f}")
print(f"  Erreur standard        : {erreur_std:.4f}")
print(f"  Statistique t calculée : {t_stat:.4f}")

# ─────────────────────────────────────────
# DÉCISION (seuil t critique ≈ 1.699 pour n=30, α=0.05)
# ─────────────────────────────────────────
print("\n" + "─" * 52)
print("3️  DÉCISION")
print("─" * 52)

t_critique = 1.699   # valeur standard pour n=30, α=0.05 unilatéral

print(f"  Valeur t calculée : {t_stat:.4f}")
print(f"  Valeur t critique : {t_critique} (seuil α = 0.05)")

if t_stat > t_critique:
    print("\n   t > t_critique → On REJETTE H0")
    print("  → La formation améliore significativement la productivité.")
else:
    print("\n   t ≤ t_critique → On NE REJETTE PAS H0")
    print("  → Pas de preuve suffisante d'amélioration.")

print("=" * 52)
```
