```python
# 1. Chargement et exploration des données

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Charger le dataset
data = pd.read_csv("train.csv")

# Aperçu
print(data.head())
print(data.shape)
print(data.info())

# Vérification des valeurs manquantes
print(data.isnull().sum())

# Statistiques descriptives
print(data.describe())
```

```python
# 2. Nettoyage et prétraitement

# Gestion des valeurs manquantes

# Remplacer ou supprimer les valeurs manquantes
data = data.dropna()  # simple stratégie
```

```python
# Encodage des variables catégorielles (si existantes)

# Exemple : conversion des variables catégorielles
categorical_cols = data.select_dtypes(include='object').columns

data = pd.get_dummies(data, columns=categorical_cols, drop_first=True)
```

```python
# Vérification finale

print(data.info())
```

```python
# 3. Analyse statistique (NumPy + SciPy)
# Séparer features et cible
X = data.drop("price_range", axis=1)
y = data["price_range"]
```

```python
# Mesures de tendance centrale
print("Mean:\n", X.mean())
print("Median:\n", X.median())
print("Mode:\n", X.mode().iloc[0])
```

```python
# Variabilité
print("Variance:\n", X.var())
print("Standard Deviation:\n", X.std())
print("Range:\n", X.max() - X.min())
```

```python
# Skewness & Kurtosis
from scipy.stats import skew, kurtosis

print("Skewness:\n", X.apply(skew))
print("Kurtosis:\n", X.apply(kurtosis))
```

```python
# Corrélation (SciPy)
corr_matrix = X.corr()

print(corr_matrix)
```

```python
# Test statistique (ANOVA entre groupes de prix)

#  on vérifie si une feature influence le prix

group0 = data[data["price_range"] == 0]["battery_power"]
group1 = data[data["price_range"] == 1]["battery_power"]
group2 = data[data["price_range"] == 2]["battery_power"]
group3 = data[data["price_range"] == 3]["battery_power"]

f_stat, p_value = stats.f_oneway(group0, group1, group2, group3)

print("F-statistic:", f_stat)
print("p-value:", p_value)
```

```python
# 4. Visualisation avec Matplotlib
#  Histogrammes
X.hist(figsize=(15, 10), bins=20)
plt.suptitle("Distribution des variables")
plt.show()
```

```python
# Boxplot (détection d’outliers)
plt.figure(figsize=(12, 6))
plt.boxplot(X["battery_power"])
plt.title("Boxplot Battery Power")
plt.show()
```

```python
# Scatter plot
plt.scatter(data["battery_power"], data["ram"], c=y, cmap="viridis")
plt.xlabel("Battery Power")
plt.ylabel("RAM")
plt.title("Relation Battery Power vs RAM")
plt.colorbar(label="Price Range")
plt.show()
```

```python
# Heatmap corrélation
import seaborn as sns

plt.figure(figsize=(12, 8))
sns.heatmap(X.corr(), annot=False, cmap="coolwarm")
plt.title("Heatmap des corrélations")
plt.show()
```

```python
# 5. Synthèse des résultats
print("Top corrélations avec price_range :")
print(data.corr()["price_range"].sort_values(ascending=False))
```
